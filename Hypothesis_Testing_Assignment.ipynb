### Null & Alternate Hypothesis

**Question 1:**
Suppose you are conducting a study to determine whether a new drug is effective in reducing blood pressure. Define the null hypothesis (H0) and the alternate hypothesis (H1) for this study. Explain the significance of each hypothesis in the context of your research.

**Answer:**
- **Null Hypothesis ($H_0$):** The new drug has no effect in reducing blood pressure (mean blood pressure change $\mu = 0$).
- **Alternate Hypothesis ($H_1$):** The new drug is effective in reducing blood pressure (mean blood pressure change $\mu < 0$).
- **Significance:** $H_0$ acts as a baseline assumption of no change, protecting against false claims. $H_1$ represents the researcher's claim that the drug works, guiding the statistical test to find evidence against $H_0$.

**Question 2:**
You are working with a manufacturing company that claims its new manufacturing process can produce car batteries with a longer lifespan than the current process. Formulate the null and alternate hypotheses for this scenario. Additionally, explain why it's essential to have clear and well-defined hypotheses before conducting hypothesis testing.

**Answer:**
- **Null Hypothesis ($H_0$):** The new process does not increase lifespan ($\mu_{new} \le \mu_{current}$). 
- **Alternate Hypothesis ($H_1$):** The new process produces batteries with a longer lifespan ($\mu_{new} > \mu_{current}$).
- **Why defined beforehand:** Clear hypotheses prevent bias, determine the appropriate statistical test (one-tailed vs. two-tailed), and establish decision boundaries before observing data.

**Question 3:**
A fast-food restaurant chain is concerned that the average delivery time for online orders has increased in recent months, potentially affecting customer satisfaction. Create the null and alternate hypotheses for testing whether there has been a significant increase in delivery times.

**Answer:**
- **Null Hypothesis ($H_0$):** The average delivery time has not increased ($\mu \le \mu_0$).
- **Alternate Hypothesis ($H_1$):** The average delivery time has significantly increased ($\mu > \mu_0$).

### Rejection Region Approach & Z-test Implementation

**Question 1:**
You are conducting a Z-test to determine whether the mean score of students in your class is significantly different from the national average. You have collected a sample of 30 student scores, and you know the population standard deviation is 10. Explain the Rejection Region Approach in the context of this test. Provide the formula for calculating the Z-score and describe how you would determine whether to reject the null hypothesis.

**Answer:**
The Rejection Region Approach defines a critical region based on the significance level ($\alpha$). If the test statistic falls into this region, $H_0$ is rejected.
Formula: $Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$
Decision: If $|Z_{calculated}| > Z_{critical}$, reject $H_0$.

```python
# Question 2: Two-sample Z-test for cholesterol reduction
np.random.seed(42)
n1, n2 = 50, 50
x1 = np.random.normal(loc=15.0, scale=3.0, size=n1)
x2 = np.random.normal(loc=12.0, scale=3.0, size=n2)

mean1, mean2 = np.mean(x1), np.mean(x2)
std1, std2 = 3.0, 3.0

pool_se = np.sqrt((std1**2 / n1) + (std2**2 / n2))
z_stat = (mean1 - mean2) / pool_se
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
# Question 3: One-sample Z-test for electric car mileage
mu_hypothesized = 250
sigma = 15.0
n = 36
sample_mean = 255.5

z_score = (sample_mean - mu_hypothesized) / (sigma / np.sqrt(n))
critical_z = stats.norm.ppf(0.99)

print(f"Calculated Z-score: {z_score:.4f}")
print(f"Critical Z-value (alpha=0.01): {critical_z:.4f}")
print("Reject H0?: ", z_score > critical_z)
```

```python
# Question 4: Z-test for difference in salaries
n_m, n_b = 100, 100
mean_m, mean_b = 85000, 72000
sigma_m, sigma_b = 8000, 6500

se_diff = np.sqrt((sigma_m**2 / n_m) + (sigma_b**2 / n_b))
z_salary = (mean_m - mean_b) / se_diff
critical_z_05 = stats.norm.ppf(0.975)

print(f"Z-score for salary difference: {z_salary:.4f}")
print(f"Critical Z (two-tailed): +/- {critical_z_05:.4f}")
```

### P-value Approach & T-test Implementation

**Case Study 1: Education**

**Assignment:**
1. **Hypotheses:** $H_0: \mu_{new} = \mu_{traditional}$ vs $H_1: \mu_{new} \neq \mu_{traditional}$
2. **T-test Implementation:** Computed via scipy.stats.

```python
np.random.seed(10)
group_new = np.random.normal(loc=78, scale=8, size=30)
group_trad = np.random.normal(loc=72, scale=8, size=30)

t_stat, p_val = stats.ttest_ind(group_new, group_trad)
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.4f}")
```

**Case Study 2: Baseball Player Performance**

**Assignment:**
1. $H_0: \mu_A \le \mu_B$ vs $H_1: \mu_A > \mu_B$
2. Conduct two-sample t-test.

```python
n1, mean1, s1 = 25, 0.285, 0.040
n2, mean2, s2 = 30, 0.270, 0.045

se = np.sqrt((s1**2 / n1) + (s2**2 / n2))
t_stat_bb = (mean1 - mean2) / se
df_num = ((s1**2 / n1) + (s2**2 / n2))**2
df_den = (((s1**2 / n1)**2) / (n1 - 1)) + (((s2**2 / n2)**2) / (n2 - 1))
df = df_num / df_den
p_val_bb = 1 - stats.t.cdf(t_stat_bb, df)

print(f"Baseball T-statistic: {t_stat_bb:.4f}")
print(f"P-value: {p_val_bb:.4f}")
```
